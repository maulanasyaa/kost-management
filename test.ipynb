```python
import jwt
from dotenv import load_dotenv
import os
```

```python
load_dotenv(".env")
```

```python
secret_jwt = os.environ.get("SECRET_JWT")
print(secret_jwt)
```

```python
payload = {"id": 7, "name": "test"}

result = jwt.encode(payload, secret_jwt, algorithm="HS256")
print(result)
```

```python
decode_result = jwt.decode(result, secret_jwt, algorithms="HS256")
print(decode_result)
```
